## Question 2: Expected Output Variable
For the road condition detection application using the RDD2022 dataset, the expected outputs are:
1. **Primary Target Variable (Classification):** `damage_classes` / `class` — Identifies the category of road defect present (e.g., `D00`, `D10`, `D20`, `D40`).
2. **Secondary Target Variable (Localization):** Spatial bounding box coordinates (`xmin`, `ymin`, `xmax`, `ymax`) specifying the precise location of defects within an image.


```python
import os
import glob
import xml.etree.ElementTree as ET
import pandas as pd

# Base path relative to your VS Code workspace directory
DATASET_PATH = "/home/topsy/3.2/AI/SemProject_MLEngine/RDD2022_sample"

# Retrieve all image paths recursively
image_files = glob.glob(os.path.join(DATASET_PATH, "**/*.jpg"), recursive=True)

data_records = []

for img_path in image_files:
    filename = os.path.basename(img_path)
    xml_path = img_path.replace('.jpg', '.xml')
    
    country = filename.split('_')[0] if '_' in filename else 'Unknown'
    width, height, depth = None, None, None
    has_damage = False
    damage_count = 0
    damage_classes = []
    
    if os.path.exists(xml_path):
        try:
            tree = ET.parse(xml_path)
            root = tree.getroot()
            
            size_node = root.find('size')
            if size_node is not None:
                width = int(size_node.find('width').text) if size_node.find('width') is not None else None
                height = int(size_node.find('height').text) if size_node.find('height') is not None else None
                depth = int(size_node.find('depth').text) if size_node.find('depth') is not None else None
            
            objects = root.findall('object')
            damage_count = len(objects)
            if damage_count > 0:
                has_damage = True
                damage_classes = list(set([obj.find('name').text for obj in objects if obj.find('name') is not None]))
        except ET.ParseError:
            pass

    data_records.append({
        'filename': filename,
        'country': country,
        'width': width,
        'height': height,
        'depth': depth,
        'has_damage': has_damage,
        'damage_count': damage_count,
        'damage_classes': ", ".join(damage_classes) if damage_classes else "None"
    })

# Primary image-level DataFrame
df = pd.DataFrame(data_records)
```
